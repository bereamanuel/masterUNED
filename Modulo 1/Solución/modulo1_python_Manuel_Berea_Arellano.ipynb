
### Cuestión 1

Se pide crear dos vectores de misma longitud con valores entre 1 y 50, siguiendo una distribución uniforme, para conseguir numeros aleatorios producidos por una distribución uniforme, utilizaremos como minimo el valor 1 y máximo 50.

```
# Definimos la longitud del vector y la semilla para poder reproducir el resultado aleatorio.
n = 10  # Longitud
seed = 17  # Semilla
m = 1  # Mínimo
M = 50  # Máximo

# Fijamos la semilla
np.random.seed(seed)

# Generamos los dos vectores pedidos
vector1 = np.random.uniform(low=m, high=M, size=n)
vector2 = np.random.uniform(low=m, high=M, size=n)

# Mostramos los resultados
print("Tenemos los dos vectores:")
print("\nVector1:")
print(vector1)
print("\nVector2:")
print(vector2)

```

Una vez generados los dos vectores se pide:

  - La suma de sus elementos
  - Productor escalar de los dos vectores
  - Correlación entre ambos

```
## Suma de sus elementos:
vector3 = vector1 + vector2
print("La suma de ambos resulta en un vector 3:")
print(vector3)
```

```
## Producto escalar
# El producto escalar es la suma del producto de ambos vectores.
productoEscalar = np.sum(vector1 * vector2)
print("Producto escalar:")
print(productoEscalar)
```

```
## Correlacion 
# Calculamos la correlación
correlacion = np.corrcoef(vector1, vector2)[0, 1]
print("Correlacion:")
print(correlacion)
```


## Cuestión 2
Se pide crear una función que dadas dos listas se encuestre los elementos comunes.

```
def interseccion(list1, list2):
    """
    Esta función toma dos listas y devuelve su intersección.
    
    INPUT:
        list1: una lista
        list2: una lista
    
    Return:
        Intersección entre list1 y list2
    
    Ejemplos:
        interseccion([1, 2, 3, 4, 5], [2, 4])
        interseccion(["a", "b", "c"], ["c", "e"])
    """
    check_cuestion2(list1)
    check_cuestion2(list2)
    
    return list(set(list1) & set(list2))

def check_cuestion2(lista):
    """
    Esta función comprueba que el input es correcto.
    """
    if not isinstance(lista, list):
        raise ValueError("Las entradas deben ser listas.")
    
    if len(lista) == 0:
        raise ValueError("La lista debe contener elementos.")
```

```
class TestInterseccion(unittest.TestCase):
    def test_numeros(self):
        self.assertEqual(interseccion([1, 2, 3, 4], [3, 4, 5, 6]), [3, 4])
    
    def test_caracteres(self):
        self.assertEqual(interseccion(["a", "b", "c"], ["b", "c", "d"]), ["b", "c"])
    
    def test_valores_repetidos(self):
        self.assertEqual(interseccion([1, 1, 2, 3], [1, 3, 4]), [1, 3])
    
    def test_sin_interseccion(self):
        self.assertEqual(interseccion([10, 20, 30], [40, 50, 60]), [])
    
    def test_listas_vacias(self):
        with self.assertRaises(ValueError):
            interseccion([], [1, 2, 3])
        with self.assertRaises(ValueError):
            interseccion([], [])
    
    def test_entradas_invalidas(self):
        with self.assertRaises(ValueError):
            interseccion("a", ["a", "B", "C"])
        with self.assertRaises(ValueError):
            interseccion([1, 2, 3], 10)


suite = unittest.TestLoader().loadTestsFromTestCase(TestInterseccion)
unittest.TextTestRunner().run(suite)
```

### Cuestión 3
Se pide crear una funcion con dos entradas, una lista de palabras y un numero entero. Debe devolver las palabras que sean iguales o mayor en longitud al numero dado.

```
def filtrado_palabras(lista, n):
    """
    Esta función toma una lista de palabras y un número entero y devuelve 
    las palabras cuya longitud sea mayor o igual al número dado.
    """
    check_cuestion3(lista, n)
    return [palabra for palabra in lista if len(palabra) >= n]

def check_cuestion3(lista, n):
    """
    Esta función comprueba que el input es correcto.
    """
    if not isinstance(lista, list) or not all(isinstance(p, str) for p in lista):
        raise ValueError("La primera entrada debe ser una lista de palabras.")
    if not isinstance(n, int):
        raise ValueError("La segunda entrada debe ser un número entero.")
```

```
class TestFiltrado(unittest.TestCase):
    
    def test_filtrado_palabras(self):
        self.assertEqual(filtrado_palabras(["Hola", "Sol", "Si"], 3), ["Hola", "Sol"])
        self.assertEqual(filtrado_palabras(["No", "Luna", "funcion"], 5), ["funcion"])
        self.assertEqual(filtrado_palabras(["Hola", "Sol"], 10), [])
    
    def test_errores_filtrado_palabras(self):
        with self.assertRaises(ValueError):
            filtrado_palabras("Hola", 3)
        with self.assertRaises(ValueError):
            filtrado_palabras(["Hola", 3, "Que"], 4)
        with self.assertRaises(ValueError):
            filtrado_palabras(["Hola", "Que", "Tal"], 5.2)
        with self.assertRaises(ValueError):
            filtrado_palabras(["Hola", "Que", "Tal"], "5")

suite = unittest.TestLoader().loadTestsFromTestCase(TestFiltrado)
unittest.TextTestRunner().run(suite)
```

## Programación básica

```
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from pyspark.sql import SparkSession

spark = SparkSession.builder \
    .appName("GymMembersAnalysis") \
    .enableHiveSupport() \
    .getOrCreate()
gym_members = spark.sql("SELECT * FROM hive_metastore.default.gym_members_exercise_tracking")
gym_members_df = gym_members.toPandas()

gym_members_df.head()
```

Se pide:

Indicar la cantidad de miembros que realizan cada tipo de entrenamiento (Workout_Type), diferenciando por nivel de experiencia (Experience_Level)

```
workout_experience_table = pd.crosstab(gym_members_df['Workout_Type'], gym_members_df['Experience_Level'])
print(workout_experience_table)
```

2. Calcular el máximo, mínimo, media y mediana de la variable Session_Duration para cada tipo de entrenamiento (Workout_Type)

```
df2 = gym_members_df.groupby('Workout_Type')['Session_Duration (hours)'].agg(
    Maximo='max',
    Minimo='min',
    Media='mean',
    Mediana='median'
).reset_index()
df2.columns = ["Workout Type", "Maximo", "Minimo", "Media", "Mediana"]
display(df2)
del df2
```

3. Convertir la variable Gender a categórica y mostrar su distribución para los diferentes tipos de entrenamiento (Workout_Type)

```
gym_members_df['Gender'] = pd.Categorical(gym_members_df['Gender'], categories=["Female", "Male"], ordered=True)
workout_gender_table = pd.crosstab(gym_members_df['Workout_Type'], gym_members_df['Gender'])
print(workout_gender_table)
```

4. Identificar los miembros que queman más de 500 calorías en una sesión y cuya frecuencia cardíaca máxima (Max_BPM) supera los 180 latidos por minuto

```
filtered_df_4 = gym_members_df[(gym_members_df['Calories_Burned'] > 500) & (gym_members_df['Max_BPM'] > 180)]
display(filtered_df_4)
```

5. 
Encontrar los miembros con menos de 1 año de experiencia (Experience_Level = 1) que no realizan entrenamientos de fuerza y tienen un porcentaje de grasa corporal (Fat_Percentage) superior al 30%

```
filtered_df_5 = gym_members_df[(gym_members_df['Experience_Level'] == 1) & 
                              (gym_members_df['Workout_Type'] != "Strength") & 
                              (gym_members_df['Fat_Percentage'] > 30)]
display(filtered_df_5)
```

6. 
Identificar los miembros que tienen una ingesta diaria de agua (Water_Intake) superior a 3 litros, realizan entrenamientos HIIT y tienen un IMC (Índice de Masa Corporal) superior a 25

```
# Calcular IMC (Índice de Masa Corporal)
gym_members_df['IMC'] = gym_members_df['Weight (kg)'] / (gym_members_df['Height (m)'] ** 2)

filtered_df_6 = gym_members_df[(gym_members_df['Water_Intake (liters)'] > 3) & 
                              (gym_members_df['Workout_Type'] == "HIIT") & 
                              (gym_members_df['IMC'] > 25)]
display(filtered_df_6)
```

7.  Crear una tabla de frecuencias cruzadas entre Workout_Frequency (categorías: <3 días/semana y ≥3 días/semana) y Experience_Level

```
gym_members_df['Workout_Frequency_group'] = gym_members_df['Workout_Frequency (days/week)'].apply(
    lambda x: "Menos 3 dias" if x < 3 else "Tres dias o mas")
workout_frequency_table = pd.crosstab(gym_members_df['Experience_Level'], gym_members_df['Workout_Frequency_group'])
print(workout_frequency_table)
```

8. Visualizar en un histograma la distribución de la variable Calories_Burned y dibujar dos líneas verticales para señalar la media (rojo) y la mediana (verde).

```
plt.hist(gym_members_df['Calories_Burned'], bins=30, edgecolor='black', alpha=0.7)
plt.title("Distribución de Calorías Quemadas")
plt.xlabel("Calorías Quemadas")
plt.ylabel("Frecuencia")

media = gym_members_df['Calories_Burned'].mean()
mediana = gym_members_df['Calories_Burned'].median()

plt.axvline(media, color='red', linestyle='dashed', linewidth=1, label='Media')
plt.axvline(mediana, color='green', linestyle='dashed', linewidth=1, label='Mediana')
plt.legend()

plt.show()
```

9.  Generar un diagrama de cajas para comparar la duración de las sesiones (Session_Duration) según el tipo de entrenamiento (Workout_Type)

```
plt.figure(figsize=(10, 6))
sns.boxplot(x='Workout_Type', y='Session_Duration (hours)', data=gym_members_df)
plt.xlabel("Tipo de entrenamiento")
plt.ylabel("Duración de la sesión en horas")
plt.title("Duración de la sesión según tipo de entrenamiento")
plt.xticks(rotation=45)
plt.show()
```
